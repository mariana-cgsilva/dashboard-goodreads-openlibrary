# Analise e processamento dos dados - Dashboard Goodreads

Este notebook documenta a etapa de preparacao dos dados usada no dashboard final. A ideia e deixar visivel o que foi feito com Pandas antes da criacao das visualizacoes em Dash.

Objetivos do processamento:

- Ler os arquivos brutos `books.csv`, `ratings.csv` e `to_read.csv`.
- Explorar estrutura, volume e qualidade inicial dos dados.
- Limpar valores ausentes, anos invalidos e categorias de idioma.
- Agregar avaliacoes e marcacoes de interesse por livro.
- Integrar as bases com `merge` usando `book_id`.
- Criar variaveis derivadas para analise: decada, autor principal, nota ponderada, proporcoes de notas e score de popularidade.
- Gerar os CSVs processados consumidos pelo dashboard.

## 1. Bibliotecas e caminhos

As bibliotecas seguem o que foi trabalhado nas aulas: Pandas para manipulacao, NumPy para calculos numericos e Plotly para graficos de apoio.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import plotly.express as px

pd.set_option("display.max_columns", 80)
pd.set_option("display.max_colwidth", 100)

BASE_DIR = Path.cwd()
RAW_DIR = BASE_DIR / "data" / "raw"
PROCESSED_DIR = BASE_DIR / "data" / "processed"

BOOKS_PATH = RAW_DIR / "books.csv"
RATINGS_PATH = RAW_DIR / "ratings.csv"
TO_READ_PATH = RAW_DIR / "to_read.csv"

for path in [BOOKS_PATH, RATINGS_PATH, TO_READ_PATH]:
    print(path.name, "->", "ok" if path.exists() else "ausente")
```

## 2. Leitura e inspecao inicial

Primeiro carregamos o catalogo de livros inteiro. Para `ratings.csv` e `to_read.csv`, que sao maiores, a agregacao sera feita em blocos mais adiante.

```python
books_raw = pd.read_csv(BOOKS_PATH)

print("Dimensoes de books.csv:", books_raw.shape)
display(books_raw.head())
display(books_raw.dtypes.to_frame("tipo"))
```

```python
def count_rows_csv(path, chunksize=1_000_000):
    total = 0
    for chunk in pd.read_csv(path, chunksize=chunksize):
        total += len(chunk)
    return total

volumes = pd.DataFrame(
    {
        "arquivo": ["books.csv", "ratings.csv", "to_read.csv"],
        "linhas": [
            len(books_raw),
            count_rows_csv(RATINGS_PATH),
            count_rows_csv(TO_READ_PATH),
        ],
    }
)
display(volumes)
```

```python
missing_summary = (
    books_raw.isna()
    .sum()
    .reset_index()
    .rename(columns={"index": "coluna", 0: "ausentes"})
)
missing_summary["percentual"] = (missing_summary["ausentes"] / len(books_raw) * 100).round(2)
display(missing_summary.sort_values("ausentes", ascending=False).head(15))
```

## 3. Funcoes de tratamento

As funcoes abaixo concentram regras de limpeza e transformacao que serao reutilizadas no pipeline.

```python
def language_group(code):
    if pd.isna(code) or str(code).strip() == "":
        return "Nao informado"

    code = str(code).strip()
    english_codes = {"eng", "en-US", "en-GB", "en-CA"}
    portuguese_codes = {"por", "pt", "pt-BR", "pt-PT"}
    spanish_codes = {"spa", "es"}
    french_codes = {"fre", "fra", "fr"}
    german_codes = {"ger", "deu", "de"}

    if code in english_codes:
        return "Ingles"
    if code in portuguese_codes:
        return "Portugues"
    if code in spanish_codes:
        return "Espanhol"
    if code in french_codes:
        return "Frances"
    if code in german_codes:
        return "Alemao"
    return "Outros"


def decade_from_year(year):
    if pd.isna(year):
        return "Nao informado"
    return f"{int(year) // 10 * 10}s"


def min_max(series):
    if series.max() == series.min():
        return series * 0
    return (series - series.min()) / (series.max() - series.min())


def score_popularidade(row):
    volume = row["ratings_count_log_norm"]
    interesse = row["to_read_count_log_norm"]
    qualidade = row["weighted_score_norm"]
    return round((0.45 * volume + 0.35 * qualidade + 0.20 * interesse) * 100, 2)
```

## 4. Limpeza do catalogo de livros

Selecionamos as colunas relevantes, convertemos campos numericos, tratamos textos ausentes e removemos anos de publicacao claramente invalidos para analise temporal.

```python
selected_columns = [
    "book_id",
    "goodreads_book_id",
    "authors",
    "original_publication_year",
    "original_title",
    "title",
    "language_code",
    "average_rating",
    "ratings_count",
    "work_ratings_count",
    "work_text_reviews_count",
    "ratings_1",
    "ratings_2",
    "ratings_3",
    "ratings_4",
    "ratings_5",
    "image_url",
    "small_image_url",
]

books = books_raw[selected_columns].copy()

numeric_columns = [
    "original_publication_year",
    "average_rating",
    "ratings_count",
    "work_ratings_count",
    "work_text_reviews_count",
    "ratings_1",
    "ratings_2",
    "ratings_3",
    "ratings_4",
    "ratings_5",
]
for column in numeric_columns:
    books[column] = pd.to_numeric(books[column], errors="coerce")

books["title"] = books["title"].fillna(books["original_title"]).fillna("Titulo nao informado")
books["authors"] = books["authors"].fillna("Autor nao informado")
books["primary_author"] = books["authors"].str.split(",").str[0].str.strip()
books["language_group"] = books["language_code"].apply(language_group)

books["publication_year"] = books["original_publication_year"].round()
books.loc[
    (books["publication_year"] < 1400) | (books["publication_year"] > 2026),
    "publication_year",
] = pd.NA
books["decade"] = books["publication_year"].apply(decade_from_year)

display(books[["book_id", "title", "primary_author", "publication_year", "language_group", "average_rating"]].head())
```

## 5. Criacao de metricas analiticas

Foram criadas variaveis para comparar qualidade, polarizacao e engajamento dos livros.

```python
books["review_rate"] = (
    books["work_text_reviews_count"] / books["work_ratings_count"].replace(0, pd.NA)
).fillna(0)

books["five_star_share"] = (
    books["ratings_5"] / books["work_ratings_count"].replace(0, pd.NA)
).fillna(0)

books["low_rating_share"] = (
    (books["ratings_1"] + books["ratings_2"]) / books["work_ratings_count"].replace(0, pd.NA)
).fillna(0)

global_mean = books["average_rating"].mean()
min_votes = 10_000
v = books["ratings_count"].fillna(0)
r = books["average_rating"].fillna(global_mean)
books["weighted_score"] = ((v / (v + min_votes)) * r + (min_votes / (v + min_votes)) * global_mean).round(3)

display(
    books[["title", "average_rating", "ratings_count", "weighted_score", "five_star_share", "low_rating_share"]]
    .sort_values("weighted_score", ascending=False)
    .head(10)
)
```

## 6. Agregacao dos arquivos grandes

`ratings.csv` e `to_read.csv` podem ser processados em blocos com `chunksize`, evitando carregar tudo de uma vez sem necessidade.

```python
book_count = pd.Series(dtype="float64")
book_sum = pd.Series(dtype="float64")
rating_distribution = pd.Series(dtype="float64")
total_rating_rows = 0

for chunk in pd.read_csv(RATINGS_PATH, chunksize=1_000_000):
    chunk["rating"] = pd.to_numeric(chunk["rating"], errors="coerce")
    chunk = chunk.dropna(subset=["book_id", "rating"])
    grouped = chunk.groupby("book_id")["rating"].agg(["count", "sum"])

    book_count = book_count.add(grouped["count"], fill_value=0)
    book_sum = book_sum.add(grouped["sum"], fill_value=0)
    rating_distribution = rating_distribution.add(chunk["rating"].value_counts(), fill_value=0)
    total_rating_rows += len(chunk)

ratings_by_book = pd.DataFrame(
    {
        "book_id": book_count.index.astype(int),
        "sample_ratings_count": book_count.astype(int).values,
        "sample_average_rating": (book_sum / book_count).round(3).values,
    }
)

ratings_distribution = rating_distribution.sort_index().rename_axis("rating").reset_index(name="count")
ratings_distribution["count"] = ratings_distribution["count"].astype(int)
ratings_distribution["percent"] = (ratings_distribution["count"] / ratings_distribution["count"].sum() * 100).round(2)

print("Linhas processadas em ratings.csv:", total_rating_rows)
display(ratings_by_book.head())
display(ratings_distribution)
```

```python
to_read_count = pd.Series(dtype="float64")
total_to_read_rows = 0

for chunk in pd.read_csv(TO_READ_PATH, chunksize=1_000_000):
    chunk = chunk.dropna(subset=["book_id"])
    grouped = chunk.groupby("book_id").size()
    to_read_count = to_read_count.add(grouped, fill_value=0)
    total_to_read_rows += len(chunk)

to_read_by_book = pd.DataFrame(
    {
        "book_id": to_read_count.index.astype(int),
        "to_read_count": to_read_count.astype(int).values,
    }
)

print("Linhas processadas em to_read.csv:", total_to_read_rows)
display(to_read_by_book.head())
```

## 7. Integracao das bases

Agora juntamos o catalogo de livros com as agregacoes de avaliacoes e interesse futuro.

```python
books = books.merge(ratings_by_book, on="book_id", how="left")
books = books.merge(to_read_by_book, on="book_id", how="left")

books["sample_ratings_count"] = books["sample_ratings_count"].fillna(0).astype(int)
books["sample_average_rating"] = books["sample_average_rating"].fillna(books["average_rating"])
books["to_read_count"] = books["to_read_count"].fillna(0).astype(int)

books["ratings_count_log_norm"] = min_max(np.log10(books["ratings_count"].clip(lower=0).fillna(0) + 1))
books["to_read_count_log_norm"] = min_max(np.log10(books["to_read_count"].clip(lower=0).fillna(0) + 1))
books["weighted_score_norm"] = min_max(books["weighted_score"])
books["popularity_score"] = books.apply(score_popularidade, axis=1)

print("Dimensoes finais:", books.shape)
display(books.head())
```

## 8. Validacao e analises de apoio

Esta etapa confere se a base processada esta coerente e gera tabelas/graficos usados para construir os insights.

```python
checks = pd.DataFrame(
    {
        "metrica": [
            "livros processados",
            "book_id duplicado",
            "titulos ausentes",
            "autores ausentes",
            "anos validos",
            "score minimo",
            "score maximo",
        ],
        "valor": [
            len(books),
            books["book_id"].duplicated().sum(),
            books["title"].isna().sum(),
            books["authors"].isna().sum(),
            books["publication_year"].notna().sum(),
            books["popularity_score"].min(),
            books["popularity_score"].max(),
        ],
    }
)
display(checks)
```

```python
top_popular = books.sort_values("popularity_score", ascending=False).head(10)
display(top_popular[["title", "primary_author", "average_rating", "ratings_count", "to_read_count", "popularity_score"]])

fig = px.bar(
    top_popular.sort_values("popularity_score"),
    x="popularity_score",
    y="title",
    orientation="h",
    color="average_rating",
    title="Top 10 livros por score combinado",
    labels={"popularity_score": "Score", "title": "Livro", "average_rating": "Nota media"},
)
fig.show()
```

```python
language_summary = (
    books.groupby("language_group", as_index=False)
    .agg(books=("book_id", "count"), avg_rating=("average_rating", "mean"), total_ratings=("ratings_count", "sum"))
    .sort_values("books", ascending=False)
)
language_summary["avg_rating"] = language_summary["avg_rating"].round(2)
display(language_summary)

px.bar(
    language_summary,
    x="language_group",
    y="books",
    color="avg_rating",
    title="Composicao da base por idioma",
    labels={"language_group": "Idioma", "books": "Livros", "avg_rating": "Nota media"},
).show()
```

```python
decade_summary = (
    books.query("decade != 'Nao informado'")
    .groupby("decade", as_index=False)
    .agg(books=("book_id", "count"), avg_rating=("average_rating", "mean"), avg_score=("popularity_score", "mean"))
    .query("books >= 20")
)
decade_summary[["avg_rating", "avg_score"]] = decade_summary[["avg_rating", "avg_score"]].round(2)
display(decade_summary.sort_values("avg_rating", ascending=False).head(10))

px.line(
    decade_summary,
    x="decade",
    y="avg_rating",
    markers=True,
    title="Media de avaliacao por decada",
    labels={"decade": "Decada", "avg_rating": "Nota media"},
).show()
```

## 9. Resumo por autor

Esta tabela apoia uma das visualizacoes do dashboard interativo.

```python
author_summary = (
    books.groupby("primary_author", as_index=False)
    .agg(
        books_count=("book_id", "count"),
        total_ratings=("ratings_count", "sum"),
        avg_rating=("average_rating", "mean"),
        avg_weighted_score=("weighted_score", "mean"),
        avg_publication_year=("publication_year", "mean"),
    )
    .query("books_count >= 2")
    .sort_values(["total_ratings", "avg_weighted_score"], ascending=False)
)
author_summary["avg_rating"] = author_summary["avg_rating"].round(2)
author_summary["avg_weighted_score"] = author_summary["avg_weighted_score"].round(2)
author_summary["avg_publication_year"] = author_summary["avg_publication_year"].round()

display(author_summary.head(15))
```

## 10. Exportacao dos dados processados

Os arquivos abaixo sao os mesmos usados pelo `app.py` para montar os dois dashboards.

```python
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

books.to_csv(PROCESSED_DIR / "books_enriched.csv", index=False)
ratings_distribution.to_csv(PROCESSED_DIR / "ratings_distribution.csv", index=False)
author_summary.to_csv(PROCESSED_DIR / "author_summary.csv", index=False)

print("Arquivos gerados:")
for path in [
    PROCESSED_DIR / "books_enriched.csv",
    PROCESSED_DIR / "ratings_distribution.csv",
    PROCESSED_DIR / "author_summary.csv",
]:
    print("-", path)
```

## 11. Conclusoes para o dashboard

- O processamento integra tres fontes e atende ao requisito de trabalhar com mais de um arquivo.
- A base final combina popularidade ja observada (`ratings_count`), qualidade percebida (`weighted_score`) e interesse futuro (`to_read_count`).
- A nota ponderada reduz distorcoes de livros com nota alta, mas poucas avaliacoes.
- As variaveis `language_group`, `decade` e `primary_author` permitem filtros e comparacoes no dashboard.
- Os CSVs exportados em `data/processed/` sao consumidos diretamente pelo dashboard em Dash.
